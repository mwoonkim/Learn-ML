```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

import numpy as np
import matplotlib.pyplot as plt

from sklearn import datasets
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score

iris = datasets.load_iris()
print(iris.feature_names)
print(iris.target_names)

data = iris.data
target = iris.target

print(data)
print(target)

plt.figure(figsize=(8, 4))
plt.boxplot([data[:,0], data[:,1], data[:,2], data[:,3]], label=iris.feature_names)
plt.title('Iris')
plt.xlabel('Features')
plt.ylabel('Length or Width (cm)')
plt.show()

X = data[:, 0]
Y = data[:, 1]
plt.scatter(X, Y, c=target)
plt.title('Iris')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.show()

scaler = StandardScaler()
data_std = scaler.fit_transform(data)
print(data_std)

plt.figure(figsize=(5, 5))
X = data_std[:, 0]
Y = data_std[:, 1]
plt.scatter(X, Y, c=target)
plt.title('Iris (Standardization)')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.show()

TWSS = {}

for k in range(1, 11):
  model = KMeans(n_clusters=k)  # 클러스터의 수를 1~11까지 차례로 fitting
  model.fit(data_std)
  TWSS[k] = model.inertia_  # 클러스터의 중심으로부터 각 데이터포인트까지의 거리(sum of square)를 계산하여 관성(inertia)이 작은 값

plt.plot(list(TWSS.keys()), list(TWSS.values()), marker='o')
plt.xlabel("Number of Clusters(K)")
plt.ylabel("TWSS")
plt.show()

K=3
kmeans = KMeans(n_clusters=K)
kmeans.fit(data_std)
centers = kmeans.cluster_centers_
labels = kmeans.labels_
print(centers)
print(labels)

plt.figure(figsize=(5, 5))
X = data_std[:, 0]
Y = data_std[:, 1]
plt.scatter(X, Y, c=target)
plt.title('Iris (Standardization)')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])

X = centers[:, 0]
Y = centers[:, 1]
plt.scatter(X, Y, c='red', s=200, alpha=0.5);
plt.show()

print(target)
print(labels)

confusion_matrix(target, labels)

for i in range(len(labels)):
  if labels[i] == 1:
    labels[i] = 0
  elif labels[i] == 0:
    labels[i] = 1

print(labels)  # 군집화에 의해 분류된 라벨

confusion_matrix(target, labels)

print(accuracy_score(target, labels, normalize=False))

print(accuracy_score(target, labels))
```
