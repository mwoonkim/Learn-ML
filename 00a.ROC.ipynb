```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
```

```python
X, y = make_classification(n_samples=1000, n_classes=2, random_state=42) # 이진 분류 데이터 생성
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42) # 데이터 세트 분리
```

```python
model = LogisticRegression() # 로지스틱 회귀 모델 학습
model.fit(X_train, y_train)
```

```python
y_scores = model.predict_proba(X_test)[:, 1] # 예측 확률 계산
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_scores) # ROC 곡선 계산
```

```python
roc_auc = auc(fpr, tpr) # AUC 계산
```

```python
plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.0])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic')
plt.legend(loc='lower right')
plt.show()
```
