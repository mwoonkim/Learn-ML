```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def init_network():
	network = { }
	network['W1'] = np.array([[0.1,0.3,0.5, 0.7],[0.2,0.4,0.6,0.8],[0.9,0.8,0.7,0.6]])
	network['B1'] = np.array([0.11,0.12,0.13,0.14])
	network['W2'] = np.array([[0.1,0.5],[0.2,0.8],[0.3,0.9],[0.4,0.8]])
	network['B2'] = np.array([0.15,0.10])

	return network
```

```python
def forward(network, x):
	W1, W2 = network['W1'], network['W2']
	b1, b2 = network['B1'], network['B2']

	a1 = np.dot(x, W1)+b1
	z1 = sigmoid(a1)
	a2 = np.dot(z1, W2)+b2

	return a2
```

```python
network = init_network()
x = np.array([1, 0.5, 0.7])
y = forward(network, x)
print(y)
```
