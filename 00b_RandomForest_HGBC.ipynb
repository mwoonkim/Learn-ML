```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, roc_auc_score
import pandas as pd

# Load iris dataset
iris_bunch = load_iris()
# Convert to DataFrame
iris_df = pd.DataFrame(data=iris_bunch.data, columns=iris_bunch.feature_names)

# target 선정
target = 'species'
# Create the binary 'species' column: 1 if the original target is 'versicolor' (which is numerical target 1), else 0.
iris_df[target] = (iris_bunch.target == 1).astype(int)

# Split data
train, test = train_test_split(iris_df, test_size = 0.2, random_state = 2)
feature = iris_df.drop(columns = target).columns
X_train_iris = train[feature] # Renamed to avoid conflict with hastie variables
y_train_iris = train[target] # Renamed to avoid conflict with hastie variables
X_test_iris = test[feature] # Renamed to avoid conflict with hastie variables
y_test_iris = test[target] # Renamed to avoid conflict with hastie variables

# RandomForest 앙상블 모델 작성
rf = RandomForestClassifier(oob_score=True, random_state=42) # Added random_state for reproducibility
rf.fit(X_train_iris, y_train_iris) # Use renamed variables

# test set에 모델 적용
y_pred_iris = rf.predict(X_test_iris) # Renamed to avoid conflict with hastie variables
y_pred_proba_iris = rf.predict_proba(X_test_iris)[:,1] # Renamed to avoid conflict with hastie variables

# test set 적용 결과
print('\n[test] accuracy:',rf.score(X_test_iris, y_test_iris))
print('[test] f1:', f1_score(y_test_iris, y_pred_iris))
print('[test] auc:', roc_auc_score(y_test_iris, y_pred_proba_iris))
print('\n oob_score:',rf.oob_score_)

# Second part: HistGradientBoostingClassifier example
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.datasets import make_hastie_10_2

# Using separate variable names to avoid overwriting the iris-related data
X_hastie, y_hastie = make_hastie_10_2(random_state=0)
X_train_hastie, X_test_hastie = X_hastie[:2000], X_hastie[2000:]
y_train_hastie, y_test_hastie = y_hastie[:2000], y_hastie[2000:]

clf = HistGradientBoostingClassifier(max_iter=100).fit(X_train_hastie, y_train_hastie)
print('\n[HistGradientBoostingClassifier] score:', clf.score(X_test_hastie, y_test_hastie)) # Added print for visibility
```
